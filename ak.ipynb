```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize, WordPunctTokenizer

nltk.download('punkt')
nltk.download('punkt_tab')

text = input("Enter your text: ")

sentences = sent_tokenize(text)
print("\n---- Sentence Tokenizer ----")
print("Tokenized sentences:", sentences)
print("Number of sentences:", len(sentences))

wp_tokenizer = WordPunctTokenizer()
wp_words = wp_tokenizer.tokenize(text)
print("\n--- Word Punct Tokenizer ---")
print("Tokenized words:", wp_words)
print("Number of tokens:", len(wp_words))
```

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize, WordPunctTokenizer, RegexpTokenizer

nltk.download('punkt')
nltk.download('punkt_tab')

text = input("Enter your text: ")

sentences = sent_tokenize(text)
print("\n---- Sentence Tokenizer ----")
print("Tokenized sentences:", sentences)
print("Number of sentences:", len(sentences))

wp_tokenizer = WordPunctTokenizer()
wp_words = wp_tokenizer.tokenize(text)
print("\n--- Word Punct Tokenizer ---")
print("Tokenized words:", wp_words)
print("Number of tokens:", len(wp_words))

regex_tokenizer = RegexpTokenizer(r'\w+')
regex_words = regex_tokenizer.tokenize(text)
print("\n---- Regexp Tokenizer ----")
print("Tokenized words:", regex_words)
print("Number of tokens:", len(regex_words))
```

```python
from nltk.tokenize import TreebankWordTokenizer

text = "Hello, world! This is a test of the Treebank Word Tokenizer."

treebank_tokenizer = TreebankWordTokenizer()

tree_words = treebank_tokenizer.tokenize(text)

print("\n---- 5. Treebank Word Tokenizer ----")
print("tokenized words:", tree_words)
print("number of items:", len(tree_words))
```
