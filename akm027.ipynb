```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('stopwords')
passage = input("enter the passage")
words = word_tokenize(passage)
stop_words = set(stopwords.words('english'))
filtered_words =[word for word in words if word.lower()not in stop_words]
print("filtered passage:"," ".join(filtered_words))
```

```python
import nltk
from nltk.corpus import wordnet
nltk.download('wordnet')
def get_synonyms_antonyms(word):
    synonyms,antonyms= set(), set()
    for synset in wordnet.synsets(word):
        synonyms.update(lemma.name()for lemma in synset.lemmas())
        antonyms.update(antonym.name()for lemma in 
        synset.lemmas() if lemma.antonyms() for antonym in lemma.antonyms())
    return{'synonyms':list(synonyms),'antonyms':
list(antonyms)}
if __name__=="__main__":
    word = input("Enter a word: ")
    results= get_synonyms_antonyms(word)
    print(f"synonyms: {results['synonyms']}")
    print(f"antonyms: {results['antonyms']}")
```

```python
import re

text = """
Hello everyone! My name is Jaliz.
Visit https://www.example.com for more information.
My email is Jaliz@gmail.com.
Follow me on Twitter @jaliz and use #NLP #Python.
I have 2 NLP projects and 3 years of Python experience.
"""
print("Original Text:")
print(text)
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
print("\nEmail Addresses:")
print(emails)
urls = re.findall(r'https?://\S+', text)
print("\nURLs:")
print(urls)
hashtags = re.findall(r'#\w+', text)
print("\nHashtags:")
print(hashtags)
mentions = re.findall(r'@\w+', text)
print("\nMentions:")
print(mentions)
numbers = re.findall(r'\b\d+\b', text)
print("\nNumbers:")
print(numbers)
clean_text = re.sub(r'https?://\S+', '', text)
clean_text = re.sub(r'\b[\w.-]+@[\w.-]+\.\w+\b', '', clean_text)
clean_text = re.sub(r'[#@]\w+', '', clean_text)
clean_text = re.sub(r'[^\w\s]', '', clean_text)
clean_text = clean_text.lower()
tokens = clean_text.split()
print("\nCleaned Text:")
print(clean_text)
print("\nTokens:")
print(tokens)
```

```python
import nltk
from nltk.corpus import words
nltk.download('words')
def find_words_starting_with(prefix):
   word_list = words.words()
   matching_words = [word for word in word_list if word.startswith(prefix)]
   return matching_words
if __name__=="__main__":
   prefix = input("Enter the prefix:")
   results = find_words_starting_with(prefix)
   print(f"Words starting with“{prefix}':{results}")
```
